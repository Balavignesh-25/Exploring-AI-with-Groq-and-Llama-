```python
# Import the Groq library for API interaction and the os module for environment variables
from groq import Groq
import os
```

```python
# Try to import 'userdata' from google.colab to determine if the environment is Google Colab
try:
    from google.colab import userdata
    IN_COLAB = True # Set flag to True if running in Colab
except ImportError:
    IN_COLAB = False # Set flag to False if not running in Colab
```

```python
def get_groq_client():
    """
    Fetches the Groq API key safely, supporting both Google Colab and local environments.
    """
    # Check if running in Google Colab to retrieve the API key from Colab Secrets
    if IN_COLAB:
        api_key = userdata.get("GROQ_API_KEY")
    # If not in Colab, retrieve the API key from environment variables
    else:
        api_key = os.getenv("GROQ_API_KEY")

    # If the API key is not found, raise a ValueError with instructions
    if not api_key:
        raise ValueError(
            "❌ GROQ_API_KEY not found.\n"
            "• In Colab → Add it in Secrets\n"
            "• Locally → Export as environment variable"
        )

    # Initialize and return the Groq client using the retrieved API key
    return Groq(api_key=api_key)
```

```python
def generate_response(
    prompt, # The user's input prompt for the AI model
    model="llama-3.1-8b-instant", # The specific AI model to use (default: Llama 3.1 8B Instant)
    temperature=0.2, # Controls the randomness of the output; lower values make output more deterministic
    max_tokens=300 # Maximum number of tokens (words/subwords) the AI should generate in its response
):
    # Obtain an authenticated Groq client instance
    client = get_groq_client()

    # Send a chat completion request to the Groq API
    response = client.chat.completions.create(
        model=model, # Specify the chosen AI model
        messages=[
            # System message to define the AI's role and behavior
            {"role": "system", "content": "You are a helpful AI assistant."},
            # User's message containing the prompt
            {"role": "user", "content": prompt}
        ],
        temperature=temperature, # Apply the specified temperature for response generation
        max_tokens=max_tokens # Limit the response to the specified maximum tokens
    )

    # Extract and return the actual content of the AI's generated message
    return response.choices[0].message.content
```

```python
# Define the prompt that will be sent to the AI model
prompt = "Explain the importance of AI in healthcare"
# Call the generate_response function with the prompt and print the AI's response
print(generate_response(prompt))
```
